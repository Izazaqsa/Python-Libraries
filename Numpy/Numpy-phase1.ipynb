# Phase 1 

```python
import numpy as np
```

### ceating an array 

```python

arr_1d = np.array([1, 2, 3, 4, 5])
print("1D array: ", arr_1d)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("2D array:\n", arr_2d)
```

### Creating array from scratch 

```python
zeros = np.zeros((3, 4))
print("zeros array: \n", zeros)

ones = np.ones((2, 3))
print("one array: \n", ones)

full = np.full((2, 2), 7)
print("full array: \n", full)

random = np.random.random((2, 3))
print("random array: \n", random)

sequence = np.arange(0, 11, 2)
print("sequnce array: \n", sequence)

array = np.arange (1,13).reshape(3,4)
print (f"array making from the aranage and reshape function : \n{array}")
```

### vector  , matrix and tensor 

```python
vector = np.array([1, 2, 3])
print("Vector: ", vector)

matrix = np.array([[1, 2, 3], 
                   [4, 5, 6]])
print("Matrix: ", matrix)

tensor = np.array([[[1, 2], [3, 4]], 
                   [[5, 6], [7, 8]]])
print("Tensor: ", tensor)
```

### array properties 

```python
arr = np.array([[1, 2, 3], 
                [4, 5, 6]])
print("Shape ", arr.shape)
print("Dimension ", arr.ndim)
print("Size ", arr.size)
print("DType ", arr.dtype)
```

### Array reshaping 

```python
arr = np.arange(12)
arr2 = np.arange (1,7).reshape (2,3)
print("Original array ", arr)

reshaped = arr.reshape((3, 4))
print("\n Reshaped array ", reshaped)

flattened = reshaped.flatten()
print("\n Flattened array ", flattened)

# ravel (returns view, instead of copy)
raveled = reshaped.ravel()
print("\n raveled array ", raveled)
print (f"the actual array (arr2)  before the ravel function: \n{arr2}")
view_array = arr2.ravel ()
print (f"\nnow applying the ravel function : \n{view_array}")
view_array [2]= 100
print (f"again printing the actual functin :\n{arr2}")


# Transpose
transpose = reshaped.T
print("\n Transposed array ", transpose)
```
