### Representing system of linear equations using matrices

for example we have a system of   {−𝑥1 + 3𝑥2 = 7, 3𝑥1 + 2𝑥2 = 1    first we will represent the coefficient of x1 and x2 and the constant to the right side in separate arrays  

```python
A = np.array(([-1,3],[3,2]), dtype=np.dtype(float))
B = np.array([7,1], dtype= np.dtype(float))
print (f"{A}\n\n{B}")
print("\n")

# applying the function 
values = np.linalg.solve(A,B)
print(f"the values for x1 and x2 are : {values}")
```

```python
# now finding the determinant of the matrix A 
determinant = np.linalg.det(A)
print(int(determinant))
```

```python
# reshaping the system 
system = np.hstack((A,B.reshape(2,1)))
print (system)
```

```python
import matplotlib.pyplot as plt 
# this is builting using the matplotlib way 
plt.plot(system[0],system[1])
plt.plot(system[1],system[0])
plt.ylabel("Y-axis")
plt.xlabel("X-axis")
plt.grid(True)
plt.legend()
plt.show()


```

System of Linear Equations
Here is a system of linear equations (or linear system) with three equations and three unknown variables:  
4𝑥1−3𝑥2+𝑥3=−10,  
2𝑥1+𝑥2+3𝑥3=0,  
−𝑥1+2𝑥2−5𝑥3=17  

To solve this system of linear equations means to find such values of the variables  𝑥1,  𝑥2,  𝑥3, that all of its equations are simultaneously satisfied.

```python
matrix_a = np.array ([[4,-3,1],
                    [2,1,3],
                    [-1,2,-5]], dtype=np.dtype(float))

matrix_b = np.array([-10,0,17],dtype=np.dtype(float))

print (f"matrix_a = {matrix_a}\nmatrix_b = {matrix_b}")
```

```python
# solving for the varaibles now by apply the linalg.solve function 
answers = np.linalg.solve(matrix_a,matrix_b)
print (f"the answers are : {answers}")
```

```python
# now finding the determinant of the matrix_a
det_a = np.linalg.det(matrix_a)
print (f"the determinant of the matrix_a is {int(det_a)}")
```
