### Drawing the rog plot 

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
mart = pd.read_excel("D:/Data Science/python/Libraries in python/Seaborn/Data sets/supermarket_sales.xlsx")
mart.head()
sub_df = mart[['Gender','Unit price','Total','Payment','gross income']]
sub_df.head()
```

```python
sns.rugplot(x='Total',data = sub_df)
```

```python
plt.Figure(figsize=(15,8))
sns.rugplot(x='Total',y='Unit price',data = sub_df,hue='Gender',height=0.1)
```

```python
plt.Figure(figsize=(15,8))
sns.kdeplot(data=mart,x='Total',y="Unit price",hue='Gender')
sns.rugplot(x='Total',y='Unit price',data = sub_df,hue='Gender',height=0.1)
```
