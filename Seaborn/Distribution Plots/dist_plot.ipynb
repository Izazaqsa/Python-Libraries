### Constructing dis plot:

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
mart = pd.read_excel("D:/Data Science/python/Libraries in python/Seaborn/Data sets/supermarket_sales.xlsx")
mart.head()
sub_df = mart[['Gender','Unit price','Total','Payment','gross income','Branch']]
sub_df.head()
```

```python
sns.displot(x='Total',data= mart, kde=True,kde_kws={'bw_adjust':0.5},rug=True,hue="Payment",multiple='stack')
```

```python
sns.displot(x='Total',data= mart, kde=True,kde_kws={'bw_adjust':0.5},rug=True,hue="Payment",multiple='stack',
            col='Branch' ,row='Gender')
```
