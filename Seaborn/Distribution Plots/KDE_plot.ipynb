### Drawing the KDE plot :

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
mart = pd.read_excel("D:/Data Science/python/Libraries in python/Seaborn/Data sets/supermarket_sales.xlsx")
mart.head()
```

```python
sub_df = mart[['Gender','Unit price','Total','Payment','gross income']]
sub_df.head()
sns.kdeplot(data= sub_df,x='Total')
```

```python
sns.kdeplot(x='Total',data= mart,bw_adjust=0.1)
# bw_adjust is the bandwidth adjustment parameter that controls the smoothness of the kernel density estimate. 
# A smaller value of bw_adjust will result in a more detailed and less smooth estimate, while a larger value 
# will produce a smoother estimate. By adjusting
```

```python
sns.kdeplot(x='Total',data= mart,bw_adjust=0.5,  hue="Payment",multiple='stack')
```
