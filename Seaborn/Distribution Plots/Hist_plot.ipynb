### Creation of Hist Plot

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
mart = pd.read_excel("D:/Data Science/python/Libraries in python/Seaborn/Data sets/supermarket_sales.xlsx")
mart.head()
```

```python
sub_df = mart[['Gender','Unit price','Total','Payment','gross income']]
sub_df.head()
```

```python
# sns.histplot(data=sub_df,x='Total')  # basics syntax for creating a hist plot 

# we can play with the number of bins and the width of the bins 
plt.figure(figsize=(15,5))
sns.histplot(data=sub_df,x='Total',bins = np.arange(0,1100,50) )
plt.xticks(np.arange(0,1100,50))

```

```python
plt.figure(figsize=(15,10))
sns.histplot(data= sub_df,x='Total',hue="Payment",bins=np.arange(0,1100,50),multiple='stack')
# multiple= stack gives the proper arrangement to the bins while using the hue parameter 
```

```python
# Bivariate histogram: A bivariate histogram shows the relationship between two variables at the same time.
sns.histplot (data=sub_df, x='Total', y='gross income')
```
