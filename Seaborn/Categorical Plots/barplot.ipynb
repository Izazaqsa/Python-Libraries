### Exploring about the barplots 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
sns.get_dataset_names() # this will give the name of the data that are in the seaborn library
dataframe = sns.load_dataset('diamonds')
dataframe.head()

# as looking at the above code we can load the inbuilt data frame in seaborn for the usage 
```

```python
path = "D:\Data Science\python\Libraries in python\Seaborn\Data sets\supermarket_sales.xlsx"
mart = pd.read_excel (path)
mart.head ()
```

```python
# now ploting the barplot 
plt.figure (figsize=(13,8))
sns.barplot(x='Product line', y= 'Total', data=mart)
```

```python
# adding hue to the bar plot to know the male and female ration 
plt.figure (figsize=(13,8))
# plt.legend()
sns.barplot(x='Product line', y= 'Total', data=mart , hue='Gender',ci=None)
# also we can change the order of the hue and also the order of the values at x-axis 
```

```python
columns = mart['Product line'].sort_values()
columns.unique()
# plotting the above boxplot in order , i mean the bar names will be in order form on the basis of the alphabets

plt.figure (figsize=(13,8))
plt.legend()
sns.barplot(x='Product line', y= 'Total', data=mart , hue='Gender',order=['Electronic_accessories',    'Fashion_accessories',     'Food_and_beverages',
      'Health_and_beauty',     'Home_and_lifestyle',      'Sports_and_travel'],hue_order=['Male','Female'],
      capsize=0.2)

# also you can set the the cap on the lines on the error bar
# if you don't want the error bar so you can remove it by setting the the ci paramater to none (ci= None)
```
