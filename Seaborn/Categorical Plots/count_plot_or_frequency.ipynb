### Learning the Seaborn count plot 

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
file_path = "D:\Data Science\python\Libraries in python\Seaborn\Data sets\supermarket_sales.xlsx"
mart = pd.read_excel(file_path)
mart.head()

# plt.figure(figsize=(15,5))
# sns.countplot(x="Product line", data= mart)
```

```python
# making it horizental , for this we will set the y in place of x 
sns.countplot(y= 'Product line', data= mart)
```

```python
# now we will know about the usage of the hue :- it gives further information about the bar , 
# like if bars tell that much of people visited at sunday then hue will tell how many of them are 
# male and females

plt.figure(figsize=(15,8))
sns.countplot(x="Product line", hue='Gender', data=mart)
```
