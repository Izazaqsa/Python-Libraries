### Cat plot:
instead of memorizing the every categorical plot separately we can use this cat plot and then we can define the desire plot that we want to draw.

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
mart = pd.read_excel("D:/Data Science/python/Libraries in python/Seaborn/Data sets/supermarket_sales.xlsx")
mart.head()
```

```python
sns.catplot(data= mart, y="Total",x='Payment',kind='violin',palette='Set2',col='Gender',row='Customer type')
```
