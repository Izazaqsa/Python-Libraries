### Creating a boxplot

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
mart = pd.read_excel ("D:/Data Science/python/Libraries in python/Seaborn/Data sets/supermarket_sales.xlsx")
mart.head()
```

```python
# creating a boxplot now 
plt.grid ()
sns.boxplot(y='Total',data= mart, width=0.2)
```

```python
colors= ['orange','skyblue','yellow']
sns.boxplot(x="Payment", y='Total', data= mart,width=0.4,palette=colors)

# in the following graph we have categories the total payment with on the basis of the payment method ,
# it shows how many of them have give them a direct cash , how many have used a credit card etc
```

```python
# in this graph we will further divide the payment into genders 
# for this we will set the hue as gender
# here we are also going to add the mean marker to the box plot and will set the parameter of it 

colors= ['orange','skyblue','yellow']
plt.grid()
sns.boxplot(x="Payment", y='Total', data= mart,width=0.4,palette=colors, hue= 'Gender',showmeans=True,meanprops={'marker':"o",
                                                                                                                 'markerfacecolor':'white',
                                                                                                                 'markeredgecolor':'black',
                                                                                                                 'markersize':'7'})
```
