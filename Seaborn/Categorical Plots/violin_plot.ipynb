### Constructing a violin plot : 
the plot is the combination of the KDE and the boxplot, where it shows that how much dense the data set it at the specific point , the fatness shows the density while the box plot inside the fatty shape gives another statistical information.
here the categrical variable will be plot on the x axis and the numerical variable will be plot on the 
y-axis

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
mart = pd.read_excel("D:/Data Science/python/Libraries in python/Seaborn/Data sets/supermarket_sales.xlsx")
mart.head ()
```

```python
sns.violinplot(y='Total', data =mart)
```

```python
sns.violinplot(x='Payment',y='Total',data= mart,hue='Gender',palette=['green','yellow'],split=True)
# the (split is used for if we have to show the hue in a single violin plot)
```

```python
# in place of the box plot in the inner shape if we have to use the line then we have to use inner parameter
sns.violinplot(x='Payment',y='Total',data= mart,hue='Gender',palette=['green','yellow'],split=True,inner="quartile")
```
