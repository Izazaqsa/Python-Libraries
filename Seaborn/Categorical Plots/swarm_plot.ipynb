### Swarm plot: 
A Swarm Plot is a categorical scatter plot where the data points are adjusted (moved slightly sideways) so that they do not overlap  
we can also apply it on the boxplot and as well as in the violin plot 

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
mart = pd.read_excel("D:/Data Science/python/Libraries in python/Seaborn/Data sets/supermarket_sales.xlsx")
mart.head()
```

```python
sns.swarmplot(y='Total',data= mart)
```

```python
# now using the categorical column also 
sns.swarmplot(data= mart, y="Total",x="Payment",palette='Set1')
```

```python
# with the additoin of the hue parameter 
sns.swarmplot(data= mart, y="Total",x="Payment",palette='Set1',hue="Gender")
```

```python
# in the above graph the gender is overlapping each other so now we have to separate it 
# creating separate one for male and also for female 

sns.swarmplot(data= mart, y="Total",x="Payment",palette='Set1',hue="Gender", dodge=True)
```
