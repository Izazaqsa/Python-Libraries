### Strip Plot : 
same like a scattor plot but it represent the occurance of the data point on a specific position through a single line or box plot or violin plot

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
mart = pd.read_excel("D:/Data Science/python/Libraries in python/Seaborn/Data sets/supermarket_sales.xlsx")
mart.head()
```

```python
sns.stripplot(y='Total',data= mart)
```

```python
sns.stripplot (y='Total',data= mart, jitter= 0.5)
# the jitter parameter is used to add some random noise to the data points, which can help to prevent
# overlapping of points in the plot, making it easier to visualize the distribution of the data.
```

```python
sns.stripplot(y="Total",data= mart, jitter = 0.5, linewidth= 1)
# with the help of it we can draw the line around the data points with specific width.
```

```python
sns.stripplot(y="Total",data= mart, x = "Payment",hue="Gender", jitter=0.1, linewidth=1,palette=["yellow","orange","pink"])
```

```python
sns.stripplot(y="Total",data= mart, x = "Payment",hue="Gender", 
              dodge=True, jitter=0.1, linewidth=1,palette=["yellow","orange","pink"])
# the dodge parameter is used to separate the data points based on the hue variable.
```

```python
# creating the strip plot over the violin plot 
sns.stripplot (y="Total", data = mart, x = "Payment",jitter = 0.2,linewidth=1)
sns.violinplot(y= "Total",data= mart,x="Payment")
```

```python
sns.stripplot (y="Total", data = mart, x = "Payment",jitter = 0.2,linewidth=1)
sns.boxplot(y='Total',data = mart,x="Payment")
```
