# Project 

```python
raw_df= pd.read_csv ("Raw_data.csv")
# print (f"the raw data frame : \n {raw_df}")
print (f'The first five rows of the raw data :\n {raw_df.head()}')
print (f"\nUnderstanding the structure of the data ")
print (raw_df.info())

# converting the date from the string to the pandas datetime object 
raw_df['Order_Date']= pd.to_datetime(raw_df["Order_Date"])

# to know how much none values there are 
print (f"\n\ncounting the none valeus :\n {raw_df.isnull ().sum()}")



# handling data 
raw_df['Customer_Name'].fillna ('Unknown', inplace=True)
raw_df['Category'].fillna('general', inplace=True)
raw_df['Quantity'].fillna (raw_df['Quantity'].min(),inplace =True)
raw_df['Price_per_unit'].fillna(raw_df['Price_per_unit'].mean(),inplace=True)
raw_df['Order_Date'].fillna (pd.Timestamp.today(), inplace =True)   # this timestamp.today() is used to load the today date 
raw_df['City'].fillna("Unknown", inplace = True)
print (f"\n\ncounting the none valeus :\n {raw_df.isnull ().sum()}")
raw_df.to_csv('Raw_data.csv')
```

## Understanding the data 

```python
print (f"Understanding the numerical statistics of the data : \n{raw_df.describe()}")

# adding column total revenue column to it
raw_df.insert (8,'Total_revenue',raw_df['Quantity']*raw_df['Price_per_unit'])
# raw_df.drop(columns='Total_revenue', inplace= True)

# print (f"after adding a new column to it : \n{raw_df}")
raw_df.to_csv('Raw_data.csv')
print (f"to know about the data :\n {raw_df.info()}")
```

## Getting insight from the data 

```python
# top selling product 
top_sells = raw_df.groupby ('Product')['Quantity'].sum()    # sum all of the product quantity 
# print (top_sells)
top_product =  top_sells.idxmax() # return the name  or the index only 
top_amount = top_sells.max ()
print (f"\nThe product name with the highest sale is {top_product} and the sold quantity is :  {top_amount}")

top_earning_categories = raw_df.groupby ('Category')['Total_revenue'].sum()
top_category = top_earning_categories.idxmax ()
top_category_amount = top_earning_categories.max ()
# print (f"top earning caterogies : {top_earning_categories}")
print (f"the name : {top_category} and the amount {top_category_amount}")

```
