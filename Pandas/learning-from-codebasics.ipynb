### Saving the data as two excel sheet in one file 

```python
# Dictionary 1
employee_data_1 = {
    "names": ["Ali", "Sara", "Ahmed", "Zara"],
    "age": [25, 30, 28, 35],
    "salary": [50000, 60000, 55000, 70000],
    "department": ["IT", "HR", "Finance", "Marketing"]
}

# Dictionary 2 
employee_data_2 = {
    "names": ["Hassan", "Ayesha", "Usman", "Fatima"],
    "age": [27, 32, 29, 40],
    "salary": [52000, 62000, 58000, 75000],
    "bonus": [5000, 7000, 6000, 8000]   
}

df1 = pd.DataFrame(employee_data_1)
df2 = pd.DataFrame(employee_data_2)

with pd.ExcelWriter("final report.xlsx") as writer :            # this is the technique that save the data as two excel sheet 
    df1.to_excel(writer,sheet_name='report1',index= False)
    df2.to_excel(writer, sheet_name='report2',index = False)

```

### using a replace function to update the data 

```python
employee_info= {
    "names": ["Ali", "Sara", "Ahmed", "Zara"],
    "age": [25, '-30C', 28, 35],        
    "salary": [50000, '60k', -55000, 70000],  
    "department": ["IT", "HR", "Finance", "Marketing"]
}

datafram= pd.DataFrame (employee_info)
print (datafram)

# using the function replace () , it need two arguments the values to be change and the values change with 

# erasing the alpabhits from the data such as 60k etc 
datafram.replace({'age':'[A-Za-z]','salary':'[A-Za-z]'}, " ",regex=True,inplace=True)   # if i did not mention the column names then it will erase all the characteristic data
print (f"\nthe new dataframe with now unit \n{datafram}")
datafram['age']= pd.to_numeric(datafram["age"], errors='coerce')
datafram['salary']=pd.to_numeric(datafram["salary"], errors='coerce')

datafram.replace([-30,-55000],np.nan,inplace = True)
print (f"\n the updated dataframe \n{datafram}")
print (f"\nfinding the null values : \n{datafram.isnull().sum()}")
print (f"\nfilling the null values")
datafram['age'].fillna(datafram["age"].mean(),inplace=True)
datafram['salary'].fillna(datafram["salary"].min(),inplace=True)
print (f"\nhandling all the data and the updated datafram : \n {datafram}")

# removing the number after the decimal 
datafram['age']=datafram['age'].round().astype(int)
print (datafram)

# no replacing the ages values with the list of values like the numeric numbers with the string 
datafram.replace([25,29,28,35],['twenty five','twenty nine','twenty eight','thirty five'],inplace=True)
print (f"\n\nthe datafram after changing the list of numeric values with the list of string values \n{datafram}")


```

## Pivot and Pivot Table 

```python
data_dict = {
    "date": ["5/1/2017", "5/2/2017", "5/3/2017", "5/1/2017", "5/2/2017", "5/3/2017", "5/1/2017", "5/2/2017", "5/3/2017"],
    "city": ["new york", "new york", "new york", "mumbai", "mumbai", "mumbai", "beijing", "beijing", "beijing"],
    "temperature": [65, 66, 68, 75, 78, 82, 80, 77, 79],
    "humidity": [56, 58, 60, 80, 83, 85, 26, 30, 35]
}


city_df = pd.DataFrame(data_dict)
print (city_df)
city_df['date']= pd.to_datetime(city_df['date'])
print (city_df.pivot (index='date',columns='city'))  # can set the values according to yourself 

# taking the aggrigate of the same values in the dataframe through the pivot table 

print (f"\nThe aggrigated values of the same rows" )
city_df.pivot_table(index='city', columns='date' , aggfunc='mean')   # by default the aggfunc  = mean , but you can set whatever you want 


# used of melt function :- used to unpivot the data of the table , actually it is need in seaborn or ml because they need the 
# long formate data not the wide formate.

weather_data = {
    "day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"],
    "chicago": [32, 30, 28, 22, 30, 20, 25],
    "chennai": [75, 77, 75, 82, 83, 81, 77],
    "berlin": [41, 43, 45, 38, 30, 45, 47]
}

weather_df = pd.DataFrame(weather_data)
print (f"the dataframe that we will apply the melt function \n")
print (weather_df)
print (f"\nafter applying the melt function")
df1 =pd.melt (weather_df,id_vars=['day'],var_name="City", value_name='Temprature')
print (f"\n{df1}")

# with the help of this we can also filter the data
print (f"\n\nAfter filtering the data \n")
print (df1[df1['City']=='chennai'])
```

## Contingency Table
shows the frequency distribution of the categorical variables at the same time

```python
data_set = {
    "Name": ["Kathy", "Linda", "Peter", "John", "Fatima", "Kadir", "Dhaval", "Sudhir", "Parvir", "Yan", "Juan", "Liang"],
    "Nationality": ["USA", "USA", "USA", "USA", "Bangladesh", "Bangladesh", "India", "India", "India", "China", "China", "China"],
    "Sex": ["Female", "Female", "Male", "Male", "Female", "Male", "Male", "Male", "Male", "Female", "Female", "Male"],
    "Age": [23, 18, 19, 22, 31, 25, 35, 31, 37, 52, 58, 43],
    "Handedness": ["Right", "Right", "Right", "Left", "Left", "Left", "Left", "Left", "Right", "Right", "Left", "Left"]
}

dataf= pd.DataFrame(data_set)
dataf
dataf1 = pd.crosstab(dataf.Handedness, dataf.Nationality)
print (dataf1)
```
