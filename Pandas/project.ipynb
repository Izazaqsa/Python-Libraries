## Combine Project in Pandas and in Numpy

```python
import pandas as pd 
import numpy as np 
```

```python
# loading the file 
data = pd.read_csv("project_data.csv")

# printing the first five row to understand the data 
print (f"First five rows of the dataframe : \n{data.head()}")

# checking the missing values ( isnull () function will be used )
print (f"\nChecking the total missing values in each column : \n{data.isnull().sum()}")

# handling the missing values 
data ['salary'].fillna (data['salary'].mean (), inplace=True)
data ['experience'].fillna (data['experience'].mean (), inplace = True)
data ['age'].fillna (data['age'].min (), inplace = True)
data['performance_rating'].fillna (data['performance_rating'].max(), inplace = True)
# print (f"\nChecking the total missing values in each column : \n{data.isnull().sum()}")   # again checking the  missing values

# finding the infinite values 

```

### detecting the inf values 

```python
numeric_col = data.select_dtypes(include=[np.number])   
print (f"finding the inifinite values \n{np.isinf(numeric_col).sum()}")
data.replace ([np.inf,-np.inf],np.nan, inplace = True) # it will replace all the inf values with the nan 
# print (f"finding the inifinite values \n{np.isinf(numeric_col).sum()}")
data.fillna (numeric_col.mean(), inplace= True) # replace the nan values with the mean of all the numeric columns 

# print (data)

# replacing negative values 
data['experience']= np.where(data['experience']< 0 , data['experience'].mean (),data['experience'])
data.to_csv("clean_project_data.csv", index = False)
```
