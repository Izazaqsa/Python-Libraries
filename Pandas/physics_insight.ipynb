```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
# 1. Load the dataset
df = pd.read_excel("physics_data.xlsx")

# Clean column names (remove leading/trailing spaces)
df.columns = df.columns.str.strip()

# 2. Extract X and Y data
# Assuming 'Field(kV/cm)' is the Electric Field (E) 
# and 'charge (µc/cm2)' is the Polarization (P)
x = df['Field(kV/cm)']
y = df['charge (µc/cm2)']

# 3. Create the plot
plt.figure(figsize=(10, 7))

# Plotting the loop with a thin line and points
plt.plot(x, y, color='red', linewidth=1, alpha=0.8, label='P-E Hysteresis Loop')
plt.scatter(x, y, s=1, color='blue', alpha=0.3) # Individual data points

# Adding axis lines for clarity (Origin)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# Formatting the graph
plt.title('P-E Hysteresis Loop', fontsize=16)
plt.xlabel('Electric Field ($E$) [kV/cm]', fontsize=14)
plt.ylabel('Polarization ($P$) [$\mu C/cm^2$]', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Display/Save the plot
plt.tight_layout()
plt.savefig('pe_hysteresis_loop.png')
```
