# Time Series Analysis

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
stock_df = pd.read_csv('Stock analysis.csv')   # this parse_dates will convert the string into date objec of the pandas 
stock_df.head ()                                                  # making the date or time as the index of the dataframe is known as datetime index 
stock_df['Date']= pd.to_datetime(stock_df['Date'])
stock_df.set_index('Date',inplace=True)
print (type(stock_df.index))
print (stock_df.loc['2025-07'])    # stock values in the july month 

stock_df.sort_index(inplace=True)

print (stock_df.loc['2025-08-15':'2025-08-25'])  # stock values in the august month , if the indices are not in sorted form then it will give an erro 
                                                 # because the slicing only work on the sorted values not on the random 


# Resampling  :- changing the frequency from one form to another , like knowing about the weekly stock except of daily 
st_df = pd.read_csv ('Daily stock.csv')

st_df['Date']= pd.to_datetime(st_df['Date'])
st_df.set_index('Date',inplace=True)
st_df.sort_index(inplace=True)
st_df.head()
print (f"\n\nIt is the range from the 10 date 20 of month september")
print (st_df.loc['2024-09-10':'2024-09-20'])

st_df.Close.resample('M').mean()

```

### Date Range 

```python
jan_st = pd.read_csv ("January stock.csv")
# produce the dates using the range function 
date = pd.date_range(start='01-01-2025',end='28-01-2025' , freq='B')  # this will produce the dates 
jan_st.set_index (date, inplace=True)    # set that dates range as the index of the dataframe 


# generating some random series or dataframe 
rng = pd.date_range(start='1/1/2025', periods=50, freq='D')
rng 
dataf = pd.DataFrame(np.random.randint (1,20,len (rng)),index=rng)   # this is the way through which we can produce a random dataframe or the series 
dataf.head ()
```

### Timestamp , Period and Period index 

```python
year = pd.Period('2025')   # yearly 
year 

month = pd.Period('2025-12', freq='M')  # monthly 
month 

# with the help of this datetime object we can perform arthemtic operations 
year += 1 
print (year )
month += 1
print (month)
print (month.start_time)    
print (month.end_time)


period_range = pd.period_range(start='2025', periods=8 , freq='M')
print (period_range)

period_df = pd.DataFrame(np.random.randint(1 , 10 , len(period_range)))
period_df.set_index(period_range)
period_df
print (period_df.index )


# exercise 
walmart_df = pd.read_csv("walmart's stock data.csv")
walmart_df.set_index('Line Item',inplace =True)
walmart_df = walmart_df.T   # take transpose on the table  
# convert the index into period index 
walmart_df.index = pd.PeriodIndex(walmart_df.index , freq='Q-JAN')  # this will conver the index into perid index , 
walmart_df["Start_time"]=walmart_df.index.map(lambda x : x.start_time)
walmart_df['End_time']= walmart_df.index.map(lambda x : x.end_time)
walmart_df
```
