### Adding columns in a dataframe
it is done by two methods , by simple assigning the column name with in a square bracket and then assign values to it and 
the other is using the insert method , this actually use to insert the column in a specific position 

```python
import pandas as pd 
```

```python
data = pd.read_csv ("output.csv")
print (f"the data of the dataframe is following : \n{data}")

# # adding with square bracket 
data ['Salary'] = [50000, 34000, 65000, 65500,70000]   
print (f"\nthe updated dataframe : \n{data}")

# adding the column with the insert function 
data.insert (0,'ID',[234,45,66,765,345])
print (f"\nagain the updated dataframe with the additioin of the id col : \n {data}")
data.to_csv("output.csv", index = False)
```

### Updating data in dataframe 
update with the use of the loc[] function , it will update a specific data , it requires two parameters , row index and the column name of the updating cell 

```python
data.loc[1,'Salary']= 80000
print (f"data after the modification of a single cell :\n{data}" )

# now updating the whole column for this we won't use the loc function 
data['Salary']= data['Salary'] + (data['Salary']*0.1)
print (f"data after increasing the salaray by 10% : \n{data}")
```

### Removing columns from the dataframe 
use of drop function df.drop(column = [column name ] , inplace = true)   the inplce = true will return the original dataframe while the inplace = false will return the new dataframe 

```python
# dropping a single column 
# data.drop (columns = ['Unnamed: 0'], inplace= True)
# print (f"dataframe after dropping the unnamed column : \n{data}")

# now dropping multiple column 
print (f"original dataframe : \n{data}")
data.drop (columns= ['ID','Salary'], inplace = True)
print (f"\ndataframe after dropping the two columns : \n {data}" )
# data.to_csv("output.csv", index = False)
```

### Handling missing data 

```python
students = {
    "name": ["Aizaz", None, "Ali", "Hira", "Zain"],
    "age": [21, None, 22, 23, 19],
    "city": ["Lahore",None, "Karachi", "Islamabad", "Multan"]
}

dataframe= pd.DataFrame(students)
print (f"students dataframe :\n{dataframe} " )

# finding the missing data 
print (f"\nfinding the missing data : \n {dataframe.isnull ()}")
# knowing the total missing data 
print (f"\ntotal number of missing data : \n {dataframe.isnull().sum()}")

# droping the missing data 
# dataframe.dropna(axis = 0 , inplace = True)   # this will drop every row which has nan values in it
# dataframe.dropna(axis = 1 , inplace = True)   # this will drop every column which has nan values in it 
# print (f"data after droping the none values \n {dataframe}")

# updating the missing values 
# dataframe.fillna (0 , inplace = True)
print (f"\ndata after filling the none values \n {dataframe}")

# filling it with calculated values 
dataframe ['name'].fillna ('Hamza',inplace = True)
dataframe ['age'].fillna (dataframe["age"].mean (), inplace = True)
print (f"\ndata after filling with calculated values the none values \n {dataframe}")
```

### Interpolation 

```python
std_info = {
    "name": ["Aizaz", "Yasir", "Ali", "Hira", "Zain"],
    "age": [21, None, 22, 23, 19],
    "city": ["Lahore","Mardan", "Karachi", "Islamabad", "Multan"]
}

info = pd.DataFrame(std_info)
print (f"the data of the students : \n{info}")

# using the interpolation :- this estimate the values from the existing values 
info.interpolate(method="linear", axis = 0 , inplace = True )
print (f"the data after using the interpolate function : \n{info}")
```

### Sorting and Aggregation 
use to arrange the data in meaningnful way , will use the sort_values () function,                                aggregation :- summarizing the numerical data / statistics  such as mean, max, min etc 

```python
data_f= pd.read_csv ("Employees data .csv")

# now sorting  a singel column 
sorted_data = data_f.sort_values(by = "Salary", ascending=False, inplace = False)
print (f"after sorting by age : \n{sorted_data}")

# # now sorting a multiple columns  
# # clear_data = data_f.sort_values(by=['Age','Salary'],ascending= [True,False],inplace= False)


# # finding the average salary 
# average_salary = data_f['Salary'].mean ()
# print (f"\nthe average salary is : {average_salary}" )

# grouping 
grouped = data_f.groupby ('Department')['Salary'].sum ()
print (f"\n{grouped}")


```

### Grouping and Joining 

```python
order  =  {'ID' : [12,34,65,76,78],
           'Name':['izaz', 'ali','yasir','ahmed','muhammad']}

order_amount = {'amount':[550, 435,645],
                'ID':[12,34,76]}

df1 = pd.DataFrame(order)
df2 = pd.DataFrame(order_amount)

# this is used to merge the dataframes , it is exactly same as the set theory , if the how  is equal to inner : it will only 
# select the rows from the two dataframes which has same values (intersection) , (outer = union) and so on  

merged_data = pd.merge(df1, df2 , on= 'ID',how='inner')
print (f"The merged data are : \n{merged_data}")


# concatenating the dataframes 

day1_data ={
    'ID': [101, 102, 103, 104, 105],
    'Name': ['Areeba', 'Sana', 'Hassan', 'Zoya', 'Tariq']
}

day2_data=  {
    'ID': [201, 202, 203, 204, 205],
    'Name': ['Iqra', 'Fahad', 'Mehwish', 'Asif', 'Rabia']
}


df_one= pd.DataFrame(day1_data)
df_two= pd.DataFrame(day2_data)
print (f"\ndataframe one : \n{df_one}")
print (f"\ndataframe two : \n{df_two}")

combine_data =pd.concat([df_one,df_two], axis = 0, ignore_index=True)
print (f"\nData after concatenation  : \n{combine_data}")

# the concat have also addition arguments which is too helpful 
concate_data = pd.concat ([df_one, df_two], keys=['day1','day2'])
print (f"\n\nthe concated data with the keys : \n {concate_data}")

# also we can print a singel dataframe merging it 
print (f"\n\nday 1 data :\n{concate_data.loc['day1']}")


```
