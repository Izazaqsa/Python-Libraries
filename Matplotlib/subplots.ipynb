# Subplots and layout adjustment

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [10,20,15,25]

plt.subplot (1,2,1) # these represent the number of rows and columns and the index number 
plt.plot (x, y)
plt.title("line chart")

plt.subplot (1,2,2)
plt.bar (x,y)
plt.title('bar chart')

plt.show ()
```

### Object oriented approach 

```python
fig, ax = plt.subplots (1,2 , figsize = (10,5))

a = [1,2,3,4]
b = [10,20,15,25]

ax [0].plot (a,b,color = 'orange')
ax[0].set_title("Line chart")

# bar chart 
ax[1].bar (a,b,color= 'green')
ax[1].set_title("bar chart")

fig.suptitle("comparison of charts ")   # use to give the title to the overall canvas 
plt.tight_layout()  # adjust the size of the graphs automatically 
plt.show ()
```

### saving the figures 

```python
x_axis = [1,2,3,4]
y_axis = [10,20,15,25]

plt.plot (x_axis, y_axis, marker = 'o', color = 'orange')
plt.xlabel ("x_axis")
plt.ylabel('y-axis')
plt.title ("student data")
plt.savefig('studentdata.pdf', dpi=300, bbox_inches = 'tight')
plt.show ()
```
