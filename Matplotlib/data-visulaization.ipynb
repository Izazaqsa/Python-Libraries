## Data visualization 

```python
import matplotlib.pyplot as plt 
```

### Bar chart

```python
product = ['sugar', 'cold drinks','vegetables','fruits']
sales = [23400, 25000, 20000,24500]

# drawing the bar chart 
plt.bar (product, sales, color = 'orange', label = 'sales 2025')
plt.xlabel('Products')
plt.ylabel("Sales")
plt.title ("Product sales graph")
plt.legend()
plt.show()
```

### Pie chart 

```python
# it is generally used to compare upto 5 or 6 types 

cities = ['peshawar','islamabad', 'lahore','karachi']
revenue = [12000,15000,10000, 20000]

plt.pie (revenue ,labels = cities, colors=['orange','skyblue','yellow','purple'],autopct='%1.1f%%')
plt.title('contribution of the cities')
plt.show()
```

### Histogram 

```python
marks = [45, 78, 62, 89, 55, 73, 91, 67, 82, 40, 59, 76, 88, 95, 53]
plt.hist(marks, bins=5 , color= "purple",edgecolor='black')
plt.xlabel("students score range")
plt.ylabel('no of studentst')
plt.title("knowing the marks")
plt.show()
```

### Scatter plot 

```python
study_hours = [2,4,6,8,10]
scores = [60,68,75,85,98]

# drawing the scatter plot 
plt.scatter(study_hours,scores , color='orange',marker='o', label="student data")
plt.xlabel ("study hours")
plt.ylabel("scores")
plt.title ("study score relationship")
plt.grid(True)
plt.legend()
plt.show ()
```
