### using pyplot functions

```python
x_axis = ['Mon','Tue','Wed','Th','Fri']
y_axis = [12,20,15,22,10]
plt.plot (x_axis, y_axis)
plt.title ("Bakery sales of the week")
plt.xlabel('Days fo the week')
plt.ylabel("Cakes sales per day")
plt.show()
```
