## Pyplot functions 

```python
import matplotlib.pyplot as plt 
```

```python
""" plt.plot (x, y , color = blue , linestyle = '--', linewidth = int value, label = ... , marker  = 'o' )
    x and y are the arguments that are needed while the other are optional """

months = [3,4,5,6]
sales = [2500, 3000, 2800, 3500]
plt.plot (months , sales , color= 'blue', linestyle='--', marker= 'o', linewidth= 2, label= '2025 sales')

# labeling 
plt.xlabel('Months')
plt.ylabel('Sales per month')

# giving a title 
plt.title("Month sales graph")

# adding descriptive box 
plt.legend(loc='upper left' , fontsize= 12)  # the parameters are optional

# adding lines at the background 
plt.grid (color='black', linestyle=':', linewidth = 1)  # parameters are optional

# to control the limits of the x axis and y axis , limit function will be used 
plt.xlim (1,8)  # the month will not be succeed of 8 
plt.ylim(1000,4000)  # also the sales will not be succeed of 4000

# using of axis ticks
plt.xticks([3,4,5,6], ['M3','M4','M5','M6'])  #replaced the 3,4,5,6 with M3 ,M4 ,M5 ,M6 also it can be used for the y axis 

plt.show()
```
