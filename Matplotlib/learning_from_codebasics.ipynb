### Plotting in a single chart 

```python
day = [1,2,3,4,5,6,7]
mix_t = [45, 47, 48 , 43, 50 , 55, 49 ]
min_t = [18, 20 , 25, 12, 15, 22, 17]
avg_t = [28,30, 32, 29, 35, 26,37]

plt.plot (day, min_t , label='Min')
plt.plot (day ,avg_t , label='Avg')
plt.plot (day, mix_t , label='Max')
plt.xlabel("Days")
plt.ylabel('Temprature')
plt.title ('Weather')
plt.legend(loc='best', shadow=True)   # the best will arrange the legend according to the free space that it fit well ,

```

### Bar chart 

```python
import numpy as np 
company = ['GOOGLE', 'MICROSOFT','FACEBOOK','NIVIDA']
revenue = [12000, 20000, 15000, 13000]
profit = [5000, 10000, 7000, 6500]

# creating a numpy array that has the same length as the company list 
xpos = np.arange(len(company))

# replacing the xpos array with the names in the graph 
plt.xticks(xpos, company)
''' i have created this np array because to show the pofit bar and the revenue bar separately , if i use the company so
it wont create the separate bars but will show them on a single bar , and then i gave the name to the np array through 
the xticks function '''


plt.bar(xpos+0.2,revenue, label='revenue', width=0.4)

# plotting the profit bar within the same graph 
plt.bar (xpos-0.2, profit, label= 'profit', width=0.4)         # actually the profit and the revenue bar are overlapping 
plt. legend()
```

### Histogram 

```python
blood_sugar_male = [118, 87 ,85, 150 , 149 , 95, 90 , 120 , 94, 135 , 137, 128]
blood_sugar_female = [120, 89, 84, 112, 115, 99, 100, 150 , 133, 88, 144,97]
plt.hist ([blood_sugar_male,blood_sugar_female], bins= [80,100,125,150],edgecolor='black', rwidth=0.95, color= ['yellow','skyblue'], label = ['male', 'female'])
plt.legend()
plt.xlabel ('Sugar level')
plt.ylabel('No of Patient')
plt.title ('Sugar analysis')
plt.show ()
```

### Pie chart 

```python
expenditure_amounts = [
    25000,  
    8000,  
    3500,    
    5000,   
    2000  
]
expenditure_items = [
    "Rent",
    "Groceries",
    "Electricity Bill",
    "Fuel",
    "Internet"
]


plt.pie (expenditure_amounts, labels=expenditure_items, autopct='%1.1f%%',radius=1.5,explode=[0,0.1,0,0,0])
# this explode will cut the pie and let the one piece to the outside 
plt.show()
```
